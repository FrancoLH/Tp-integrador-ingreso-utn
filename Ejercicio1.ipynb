# 1. Una motocicleta que se encuentra inicialmente en reposo, acelera constantemente a 5 m/s2. 
 ### a) Determine la distancia que recorre la motocicleta al cabo de 4 s.  
 ### b) Grafica, usando la Liberia de Python, 
    1) la velocidad del móvil en función del tiempo
    2) la posición de la motocicleta en función del tiempo 
    3) la aceleración en función del tiempo. 

## Punto 1.a


$$x_t = x_0 + V(t - t_0)$$

```python
# Aceleracion
a = 5
#Velocidad inicial
v0 = 0
# Ubicacion inicial
x0 = 0  
# Tiempo inicial y final
t0 = 0
tf = 4
# Cálculo de la posición final
xf = x0 + v0 * (tf - t0) + 0.5 * a * (tf - t0) ** 2

print("La posicion final de la moto es de:", xf,"metros")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Datos definidos en la primera celda
v0 = 0  # La moto parte del reposo
tiempo = np.linspace(t0, tf, 100)

# Ecuaciones del movimiento rectilineo uniforme
velocidad = v0 + a * (tiempo - t0)
posicion = x0 + v0 * (tiempo - t0) + 0.5 * a * (tiempo - t0) ** 2
aceleracion = np.full_like(tiempo, a)

# Graficar velocidad, posicion y aceleracion
fig, graficos = plt.subplots(3, 1, figsize=(8, 10), sharex=True)

graficos[0].plot(tiempo, velocidad, color="royalblue")
graficos[0].set_title("Velocidad en funcion del tiempo")
graficos[0].set_ylabel("Velocidad (m/s)")
graficos[0].grid(True)

graficos[1].plot(tiempo, posicion, color="seagreen")
graficos[1].set_title("Posicion en funcion del tiempo")
graficos[1].set_ylabel("Posicion (m)")
graficos[1].grid(True)

graficos[2].plot(tiempo, aceleracion, color="firebrick")
graficos[2].set_title("Aceleracion en funcion del tiempo")
graficos[2].set_xlabel("Tiempo (s)")
graficos[2].set_ylabel("Aceleracion (m/s2)")
graficos[2].grid(True)

plt.tight_layout()
plt.show()
```
