 # 2. Un móvil, que puede considerarse una partícula, comienza a observarse en el instante en el que su rapidez es de 8m/s. Si acelera uniformemente su marcha en la misma dirección y sentido en el que se está moviendo, de forma que recorre 640m en 40s: 
## a) Determine la aceleración del movimiento. 
## b) Calcule la velocidad media en estos 40s. 
## c) Calcule la rapidez final. 
## d) Calcular la aceleración a en m/s2

# PUNTO 2.a
$$d = v_i \cdot t + \frac{1}{2} a \cdot t^2$$   $$a = \frac{2 \cdot (d - (v_i \cdot t))}{t^2}$$

```python
import numpy as np
import matplotlib.pyplot as plt

Vi = 8
distancia = 640
tiempo = 40
aceleracion = 2 * (distancia - (Vi * tiempo)) / (tiempo ** 2)
print ("la aceleracion es:", aceleracion)
```

# PUNTO 2.b
$$v_m = \frac{d}{t}$$

```python
Vm = distancia / tiempo
print ("la velocidad media es:", Vm, "m/s²")
```

# PUNTO 2.c
$$v_f = v_i + a \cdot t$$

```python
Vf = Vi + (aceleracion * tiempo)
print ("la velocidad final es:", Vf, "m/s²")
```

# PUNTO 2.d

```python
aceleracion = 2 * (distancia - (Vi * tiempo)) / (tiempo ** 2)
print ("la aceleracion es:", aceleracion, "m/s²")
```
