# 3. Un vehículo que se mueve en línea recta con una rapidez de 15 m/s disminuye su rapidez a razón de 1 m/s cada segundo. 

## a) Determine el tiempo que tardará en detenerse. 
## b) Grafique, usando la Liberia de Python, la velocidad del móvil en función del tiempo y la posición del móvil en función del tiempo.  

# PUNTO 3.a
$$v_f = v_i + a \cdot t$$ $$t = \frac{v_f - v_i}{a}$$

```python
import numpy as np
import matplotlib.pyplot as plt

#variables iniciales
Vi = 15
aceleracion = -1
Vf = 0

#cálculo tiempo inicial
tiempo = (Vf - Vi) / aceleracion
print("el tiempo es:", tiempo, "s")
```

# PUNTO 3.b
$$d = v_i \cdot t + \frac{1}{2} \cdot a \cdot t^2$$

```python
#100 puntos de tiempo desde el segundo 0s hasta 15s
tiempo = np.linspace(0, tiempo, 100)

#velocidad y posición para cada instante t
v = Vi + aceleracion * tiempo
d = Vi * tiempo + 0.5 * aceleracion * (tiempo**2)

#12 de ancho por 5 de alto
plt.figure(figsize=(12, 5))

# Gráfico 1: Velocidad
plt.subplot(1, 2, 1)
plt.plot(tiempo, v, color='red', linewidth=2)
plt.title('Velocidad en función del tiempo')
plt.xlabel('Tiempo (s)')
plt.ylabel('Velocidad (m/s)')
plt.grid(True)

# Gráfico 2: Posición
plt.subplot(1, 2, 2)
plt.plot(tiempo, d, color='blue', linewidth=2)
plt.title('Posición en función del tiempo')
plt.xlabel('Tiempo (s)')
plt.ylabel('Posición (m)')
plt.grid(True)


plt.tight_layout()
plt.show()
```
